```python
print("database shape:",df.shape)
print("\n first 5 rows:",df.head(5))

print("\n Missing values:",df.isnull().sum())
```

```python
# step1: input and output columns

x = df["Open"].values
y= df["Close"].values

x_mean = np.mean(x)
x_std = np.std(x)
x_normalized = (x - x_mean) / x_std

m=0
c=0
learning_rate = 0.01
epochs = 5000
n = len(x_normalized)

for i in range(epochs):
    # calculate predicted close price
    y_pred = (
        m*x_normalized + c
    )
    # calculate predict error
    error = (
        y_pred - y
    )
    # calculate gradient for slope
    gradient_m = (
        2 / n
    ) * np.sum(
        error * x_normalized
    )
    # calculate gradient for intercept
    gradient_c = (
        2 / n
    ) * np.sum(
        error
    )
    #update slope
    m = (
        m - learning_rate * gradient_m
    )
    #update intercept
    c = (
        c - learning_rate * gradient_c
    )

    
predicted_y = (
    m * x_normalized + c        
)

mse = np.mean(
    (y - predicted_y) ** 2
)


total_error = np.sum(
    (y - np.mean(y)) ** 2
)
remaining_error = np.sum(
    (y - predicted_y) ** 2
)
r_squared = (
    1 - remaining_error / total_error
)


print("\n================ RESULTS ================\n")
print("Slope(m):", m)
print("Intercept(c):", c)
print("Mean Squared Error:", mse)
print("R-Squared Score:", r_squared)


result = pd.DataFrame({
    "Open Price" : x,
    "Actual Close Price" : y,
    "Predicted Close Price" : predicted_y
})
print("\n Actual and Predicted Values:",result.head(10))


sorted_indexes = np.argsort(x)
plt.figure(figsize=(12,6))
plt.scatter(x,y,alpha=0.4,label='Actual Closing Price')
plt.plot(x[sorted_indexes] , predicted_y[sorted_indexes] , linewidth = 2 , label= 'Predicted Regression Line')
plt.xlabel("Tesla opening price")
plt.ylabel("Tesla Closing price")
plt.title("Tesla stock price prediction" "Simple Linear Regression")
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.model_selection import TimeSeriesSplit
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import (
    RandomForestRegressor,
    AdaBoostRegressor,
    GradientBoostingRegressor
)
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date")
df = df.reset_index(drop = True)

print("\n Data after sort:")
print(df.head())
```

```python
## Handle missing values

print("\n Missing value before :")
print(df.isnull().sum())

df = df.dropna()
print("\n Missing value after :")
print(df.isnull().sum())
```

```python
## Select features and target
Features = [
    "Open",
    "High",
    "Low",
    "Volume"
]
target = "Close"
x = df[Features]
y = df[target]

print("\n Features:")
print(x.head())

print("\n Targets:")
print(y.head())
```

```python
## Train_test_split
split_index = int(len(df) * 0.80 )

x_train = x.iloc[:split_index]
x_test = x.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("\n Training Data size:")
print(x_train.shape)

print("\n Testing Data size:")
print(x_test.shape)
```

```python
## Evaluation fn
def evaluate_model(model,x_train,x_test,y_train,y_test,name):
    model.fit(x_train,y_train)
    train_pred = model.predict(x_train)
    test_pred = model.predict(x_test)

    ## Train_metrics
    train_rss = np.sum(
        (y_train - train_pred) ** 2
    )
    train_rmse = np.sqrt(
        mean_squared_error(y_train,train_pred)
    )
    train_r2 = r2_score(
        y_train,
        train_pred
    )

    ## Test_metrics
    test_rss = np.sum(
        (y_test - test_pred) ** 2
    )
    test_rmse = np.sqrt(
        mean_squared_error(y_test,test_pred)
    )
    test_r2 = r2_score(
        y_test,
        test_pred
    )
    test_mae = mean_absolute_error(
        y_test,
        test_pred
    )

    ## print
    print("\n"+ name + ":")

    print("\nTRAINING RESULTS")
    print("RSS  :", train_rss)
    print("RMSE :", train_rmse)
    print("R2   :", train_r2)

    print("\nTESTING RESULTS")
    print("RSS  :", test_rss)
    print("RMSE :", test_rmse)
    print("R2   :", test_r2)
    print("MAE  :", test_mae)


    ## overfit / underfit
    print("\nFIT CHECK")
    r2_diff = train_r2 - test_r2

    if r2_diff >0.15:
        print("Possible Overfitting")
        print("Training R2 is considerably higher than Testing R2.")
    
    elif train_r2 < 0.50 and test_r2 < 0.5 :
        print("Possible Underfitting")
        print("Both training and testing R2 are relatively low.")

    else:
        print("No strong indication of overfitting/underfitting")
        print("Training and testing performance are reasonably comparable.")

    return {
        "Model": name,
        "Train RSS": train_rss,
        "Test RSS": test_rss,
        "Train RMSE": train_rmse,
        "Test RMSE": test_rmse,
        "Train R2": train_r2,
        "Test R2": test_r2,
        "MAE": test_mae,
        "Model Object": model
    }
```

```python
## Linear Regression
linear_model = LinearRegression()
result = evaluate_model(
    linear_model,
    x_train,
    x_test,
    y_train,
    y_test,
    "Linear Regression"
)
```

```python
## Random Forest
random_forest = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)
rf_result = evaluate_model(
    random_forest,
    x_train,
    x_test,
    y_train,
    y_test,
    "Random Forest"
)
```

```python
## AdaBoost
ada_model = AdaBoostRegressor(
    n_estimators=200,
    learning_rate=0.05,
    random_state=42
)
ada_result = evaluate_model(
    ada_model,
    x_train,
    x_test,
    y_train,
    y_test,
    "AdaBoost"
)
```

```python
## Gradient Boosting
gradient_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)
gradient_result = evaluate_model(
    gradient_model,
    x_train,
    x_test,
    y_train,
    y_test,
    "Gradient Boosting"
)
```

```python
## support vector 
svr_model = Pipeline([
    ("scalar", StandardScaler()),
    ("SVR", SVR(
        kernel="rbf",
        C= 100,
        gamma= "scale",
        epsilon= 0.1
    ))
])

svr_result = evaluate_model(
    svr_model,
    x_train,
    x_test,
    y_train,
    y_test,
    "SVR"
)
```

```python
## Store all models
store_result = [
    result,
    rf_result,
    ada_result,
    gradient_result,
    svr_result
]
comparison_df = pd.DataFrame(store_result)

display_columns = [
    "Model",
    "Train RSS",
    "Test RSS",
    "Train RMSE",
    "Test RMSE",
    "Train R2",
    "Test R2",
    "MAE"
]
print("\n Model Comparison:")
print(comparison_df[display_columns].round(4))
```

```python
## 5-Fold validation
print("\n")
print("5-FOLD TIME-SERIES CROSS-VALIDATION")

tscv  = TimeSeriesSplit(n_splits=5)

## function for CV 
def cross_validate_model(model,x,y,model_name):
    fold_scores = []
    fold_number = 1

    for train_index,validation_index in tscv.split(x):
        x_cv_train = x.iloc[train_index]
        x_cv_test = x.iloc[validation_index]

        y_cv_train = y.iloc[train_index]
        y_cv_test = y.iloc[validation_index]

        model.fit(x_cv_train,y_cv_train)
        predictions = model.predict(x_cv_test)

        rmse = np.sqrt(
            mean_squared_error(
                y_cv_test,
                predictions
            )
        )

        fold_scores.append(rmse)
        print(model_name,"-Fold", fold_number,"RMSE:",round(rmse,4))
        fold_number += 1
    
    avg_rmse = np.mean(fold_scores)
    std_rmse = np.std(fold_scores)

    print("\n Avg RMSE:", round(avg_rmse,4))
    print("RMSE Std Deviation:", round(std_rmse,4))

    return {
        "Model" : model_name,
        "CV Avg RMSE" : avg_rmse,
        "CV Std RMSE" : std_rmse
    }
```

```python
## Cross- Validate models
cv_result = []
cv_result.append(
    cross_validate_model(
        LinearRegression(),
        x_train,
        y_train,
        "Linear Regression"
    )
)

cv_result.append(
    cross_validate_model(
        RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ),
        x_train,
        y_train,
        "Random Forest"
    )
)

cv_result.append(
    cross_validate_model(
        AdaBoostRegressor(
            n_estimators=200,
            learning_rate=0.05,
            random_state=42
        ),
        x_train,
        y_train,
        "AdaBoost"
    )
)

cv_result.append(
    cross_validate_model(
        GradientBoostingRegressor(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ),
        x_train,
        y_train,
        "Gradient Boosting"
    )
)

cv_result.append(
    cross_validate_model(
        Pipeline([
            ("scaler", StandardScaler()),
            ("svr", SVR(
                kernel="rbf",
                C=100,
                gamma="scale",
                epsilon=0.1
            ))
        ]),
        x_train,
        y_train,
        "SVR"
    )
)
```

```python
## Cross Validation RMSE For Each Fold
print("\nGenerating cross-validation fold graph...")

cv_fold_results = []

## Function to calc
def get_fold_rmse(model, x, y):

    fold_scores = []

    for train_index, validation_index in tscv.split(x):

        x_cv_train = x.iloc[train_index]
        x_cv_test = x.iloc[validation_index]

        y_cv_train = y.iloc[train_index]
        y_cv_test = y.iloc[validation_index]

        model.fit(
            x_cv_train,
            y_cv_train
        )

        predictions = model.predict(
            x_cv_test
        )

        rmse = np.sqrt(
            mean_squared_error(
                y_cv_test,
                predictions
            )
        )

        fold_scores.append(rmse)

    return fold_scores


## Models For CV Graph
cv_models = {

    "Linear Regression":
        LinearRegression(),

    "Random Forest":
        RandomForestRegressor(
            n_estimators=200,
            random_state=42,
            n_jobs=-1
        ),

    "AdaBoost":
        AdaBoostRegressor(
            n_estimators=200,
            learning_rate=0.05,
            random_state=42
        ),

    "Gradient Boosting":
        GradientBoostingRegressor(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        ),

    "SVR":
        Pipeline([
            (
                "scaler",
                StandardScaler()
            ),
            (
                "svr",
                SVR(
                    kernel="rbf",
                    C=100,
                    gamma="scale",
                    epsilon=0.1
                )
            )
        ])
}


for model_name, model in cv_models.items():

    scores = get_fold_rmse(
        model,
        x_train,
        y_train
    )

    cv_fold_results.append({
        "Model": model_name,
        "Fold 1": scores[0],
        "Fold 2": scores[1],
        "Fold 3": scores[2],
        "Fold 4": scores[3],
        "Fold 5": scores[4]
    })


cv_fold_df = pd.DataFrame(
    cv_fold_results
)

## plot each model CV performacne
plt.figure(figsize=(12, 7))

fold_numbers = [1,2,3,4,5]

for index, row in cv_fold_df.iterrows():

    scores = [
        row["Fold 1"],
        row["Fold 2"],
        row["Fold 3"],
        row["Fold 4"],
        row["Fold 5"]
    ]

    plt.plot(
        fold_numbers,
        scores,
        marker="o",
        label=row["Model"]
    )


plt.title(
    "5-Fold Time-Series Cross-Validation RMSE"
)

plt.xlabel("Fold")
plt.ylabel("RMSE")

plt.xticks(
    fold_numbers
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()
plt.show()
```

```python

# AVERAGE CROSS-VALIDATION RMSE
plt.figure(figsize=(12, 6))

plt.bar(
    cv_df["Model"],
    cv_df["CV Avg RMSE"]
)

plt.title(
    "Average Cross-Validation RMSE"
)

plt.xlabel("Model")

plt.ylabel(
    "Average CV RMSE"
)

plt.xticks(
    rotation=45,
    ha="right"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()
```
